```python
import math

x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

d = math.sqrt(math.pow(x2 - x1, 2) + math.pow (y2 - y1, 2))

print ("The distance is:", d)


#1. The math library helped simplify my program by providing built-in functions for mathematical calculations. The math library also helped make the code shorter, easier, and more understandable

#2. The sqrt () function made it easy to find the square root. Another function is the pow() function since it made it easy to square the differences between the coordinates.

#3. Without sqrt() and pow(), I wuld have to write the calculations manually which would result to a longer program. It might also lead tomore errors and comfusions fotr me.
```
